1. create a 1D numpy array named temperatures_w1 for week 1 with the values [22.5,25.3,20.8,23.4,26.1,24.8,21.9]

```python
import numpy as np
```

```python
temperatures_w1=np.array([22.5,25.3,20.8,23.4,26.1,24.8,21.9])
print(temperatures_w1)
```

```python
print('shape:',temperatures_w1.shape)
print('datatype:',temperatures_w1.dtype)
print('number of elements:',temperatures_w1.size)
```

```python
fahrenheit=(temperatures_w1 * 9/5)+32
print('temperature in fahrenheit:',fahrenheit)
```

```python
print('maximum:',np.max(temperatures_w1))
print('minimum:',np.min(temperatures_w1))
print('mean:',np.mean(temperatures_w1))
```

PANDAS SERIES

1.Creating pandas series

create a pandas series named marks with the following values

marks: 95,92,89,85,80

custom indices:rank1,rank2,rank3,rank4,rank5

pd.Series() create the series index= assigns custom labels to each mark

```python
import pandas as pd
```

```python
marks=pd.Series([95,92,89,85,80],index=['Rank1','Rank2','Rank3','Rank4','Rank5'])
print(marks)
```

2.indexing and slicing



*  use the integer index position to access the marks of first 1st rank student.
print(marks.iloc[0]) .iloc access by integer position

* use loc accessor to retrieve the marks of the top 3 ranks by specifying their index label

print(marks.loc[['Rank1','Rank2','Rank3']] .loc access by index label


*  use iloc accessor to retrieve the mark of 3rd rank student
print(marks.iloc[2])

*   apply a boolean mask to filter the series and retrieve the ranks where marks are greater than 90
print(marks[marks>90])

```python
print(marks.iloc[0])
print(marks.loc[['Rank1','Rank2','Rank3']])
print(marks.iloc[2])
print(marks[marks>90])
```

3.manipulating series

modify the mark of the student with the first rank to 100

marks.loc['Rank1']=100 updates the value of rank1

```python
marks.loc['Rank1']=100
print(marks)
```

remove the entry corresponding to the last rank from the series

.drop() removes the entry

```python
marks=marks.drop('Rank5')
print(marks)
```

compute the cgpa by dividing each mark in the series by 10


```python
cgpa=marks/10
print(cgpa)
```

PANDAS DATA FRAME

1.CREATING PANDAS DATA FRAME

create a pandas data frame named transactions with the following values

```python
transactions=pd.DataFrame({'TransactionID':[101,102,103,104,105,106,107,108,109,110],
                           'ProductCategory':['Electronics','Clothing','Electronics','Furniture','Clothing','Electronics','Furniture',
                                              'Clothing','Furniture','Electronics'],
                           'Region':['North','South','North','East','West','North','East','West','South','North'],
                           'Amount':[200,150,300,450,200,250,300,180,350,400]})
print(transactions)
```

2.data exploration

display the transactions dataframe and its basic informations including head,tail,shape,column name and datatype


*  head() displays the first 5 rows
*  tail() displays the last 5 rows





```python
print(transactions)
```

```python
print(transactions.head())
```

```python
print(transactions.tail())
```



* shape returns(10,4) means 10 rows and 4 columns
*  columns display the column names

*  info() shows column datatype and number of non null values






```python
print(transactions.shape)
print(transactions.columns)
print(transactions.info())
```

display only the productcategory and amount column

double square bracets used to select two columns

```python
print(transactions[['ProductCategory','Amount']])
```

retrieve the last 3 columns of the dataframe using iloc or loc

.iloc[:,-3:] selects all rows and last 3 columns

```python
print(transactions.iloc[:,-3:])
```

filter rows where the region is north and amount is greater than 200
transactions['Region']=='North' checks the region

 transactions['Amount']>200 checks the amount

  & means both conclusion must be true

```python
result=transactions[(transactions['Region']=='North') & (transactions['Amount']>200)]
print(result)
```

find the value counts for the product category column

value_counts() counts how many time each category appears

```python
print(transactions['ProductCategory'].value_counts())
```

find the unique values in the region column

unique() returns each distinct region

```python
print(transactions['Region'].unique())
```

group by region anf find the mean amount for each region

groupby('Region') GROUPS ROWS THAT HAVE SAME REGION
['Amount'].mean() calculates the average amount for each region

```python
print(transactions.groupby('Region')['Amount'].mean())
```

3.manipulating the data frame

modify the amount for the transaction with transaction id 102 to 165

```python
transactions.loc[transactions['TransactionID']==102,'Amount']=165
print(transactions)
```

add a new column discount by calculating 10% of the amount

```python
transactions['Discount']=transactions['Amount']*0.10
print(transactions)
```

remove the row with transactionid 109

```python
transactions=transactions[transactions['TransactionID']!=109]
print(transactions)
```

delete the discount column from the dataframe

```python
transactions=transactions.drop(columns=['Discount'])
print(transactions)
```
